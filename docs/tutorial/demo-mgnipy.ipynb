# `MGnipy` and `ABaCo` demo: fetching and integrating studies from MGnify API

```python
# ignore debugging warning
import logging
logging.basicConfig(level=logging.WARNING)
# import packages
from mgnipy import MGnipy
```

```python
MG = MGnipy()

studies = MG.studies(
    search="colorectal cancer"
)

studies.bulk_fetch()

studies.enrich_details()

studies.to_df(expand_nested_dicts=True)
```

```python
MZ = studies.datasets

print(MZ)

MZ.downloads_df()
```

```python
ssu = MZ["Taxonomic assignments SSU"]

print(type(ssu))
print(ssu)

ssu.downloads_df()
```

```python
ssu.load()

ssu.to_pandas()
```

```python
ssu.enrich_runs()
```

```python
adata=ssu.to_anndata()
```
